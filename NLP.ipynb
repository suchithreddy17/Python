```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt_tab')
nltk.download('stopwords')

text = "Natural Language Processing makes computers understand human language."

words = word_tokenize(text)
print("Tokenized Words:", words)

stop_words = set(stopwords.words("english"))
filtered_words = [w for w in words if w.lower() not in stop_words]

print("After Removing Stop Words:", filtered_words)

```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

ps = PorterStemmer()
words = ["playing", "played", "plays", "happily", "running"]

for w in words:
    print(w, "→", ps.stem(w))
```

```python
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

text = "Machine learning is fun. NLP is a part of machine learning."

words = word_tokenize(text)
fd = FreqDist(words)

print(fd.most_common())
```

```python
import nltk
from nltk.util import bigrams
from nltk.tokenize import word_tokenize

text = "NLP is fun and NLP is interesting"
tokens = word_tokenize(text)

bigrams_list = list(bigrams(tokens))
print(bigrams_list)
```

```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize
import nltk
nltk.download('wordnet')

sentence = "I went to the bank to deposit money."
word = "bank"

sense = lesk(word_tokenize(sentence), word)
print("Meaning:", sense)
print("Definition:", sense.definition())

```

```python
import nltk
from nltk.stem import PorterStemmer

nltk.download('punkt')

ps = PorterStemmer()
words = nltk.word_tokenize("The children are playing happily in the playground.")

stems = [ps.stem(w) for w in words]
print(stems)
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('averaged_perceptron_tagger_eng')
sentence = "The quick brown fox jumps over the lazy dog."

words = word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

print(pos_tags)
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars"]
for w in words:
    print(w, "→", lemmatizer.lemmatize(w, pos='v'))
```

```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize

text = "NLP makes computers understand language"
tokens = word_tokenize(text)

print("Bigrams:", list(ngrams(tokens, 2)))
print("Trigrams:", list(ngrams(tokens, 3)))
```

```python
from collections import Counter

sentence = "the cat sat on the mat"
tokens = sentence.split()

# Bigram counts
bigrams = [(tokens[i], tokens[i+1]) for i in range(len(tokens)-1)]
bigram_counts = Counter(bigrams)

# Laplace smoothing
def laplace_prob(word1, word2):
    return (bigram_counts[(word1, word2)] + 1) / (len(bigrams) + len(set(tokens)))

print("P(cat | the) =", laplace_prob("the", "cat"))
```

```python
import pyttsx3
from IPython.display import Audio, display
import os

# Install eSpeak-ng for pyttsx3 to work
# Initialize text-to-speech engine
engine = pyttsx3.init()

# Take manual input from user
text = input("Enter text to convert into voice: ")

# Set the output file name
output_filename = "output.wav"

# Save the speech to a file instead of playing directly
engine.save_to_file(text, output_filename)
engine.runAndWait()

# Play the saved audio file in Colab
if os.path.exists(output_filename):
    print(f"Playing audio for: '{text}'")
    display(Audio(output_filename, autoplay=True))
else:
    print("Error: Audio file not created.")
```
